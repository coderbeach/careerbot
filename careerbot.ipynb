```python

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen2.5-0.5B-Instruct"

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Running on:", device)

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto"
).to(device)

print("CareerCompass AI model loaded!")
```

```python

SYSTEM_PROMPT = """
You are CareerCompass AI, a friendly and practical career
guidance assistant for students and beginners.

Your responsibilities:
1. Explain different career paths and job roles.
2. Recommend skills and learning roadmaps.
3. Suggest beginner-friendly projects.
4. Compare careers based on skills, interests and goals.
5. Ask follow-up questions when the user's goals are unclear.

Give clear, structured, actionable answers.
Do not guarantee jobs, salaries or career outcomes.
Be honest when information is uncertain.
If you do not know something, say so.
"""

chat_history = [
    {"role": "system", "content": SYSTEM_PROMPT}
]

def chat(user_message):
    chat_history.append({
        "role": "user",
        "content": user_message
    })

    prompt = tokenizer.apply_chat_template(
        chat_history,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=450,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            pad_token_id=tokenizer.eos_token_id
        )

    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]
    answer = tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()

    chat_history.append({
        "role": "assistant",
        "content": answer
    })

    return answer

print(chat("I am a CSE student interested in artificial intelligence."))
```
